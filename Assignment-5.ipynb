```python
#  2) Write a Python program to join three separate tuples into one new tuple using the + operator.

t1 = ('a', 'b', 'c')
t2 = (3, 6, 7)
t3 = ('mango', 'orange', 'banana')

t4 = t1 + t2 + t3

print(t4)
```

```python
# 3) Write a Python program to check whether a specific element exists inside a tuple using the in keyword.

t = ('mango', 'orange', 'banana')

if 'orange'in t:
    print("orange exists in tuple")
else:
    print("orange does not exist in tuple")
```

```python
# 6) Write a Python program to determine how many elements are in a set without using the built-in len() function. s = {"cat", "dog", "bird", "fish"}

s = {"cat", "dog", "bird", "fish"}

n = 0
for i in s:
    n = n+1

print("Total number of elements in tuple 's' are", n)
```

```python
#  7) Write a Python program to combine two sets into one, containing all unique elements from both sets.

s1 = {2, 3, 4,7}
s2 = {6, 3, 8, 4}

s3 = s1.union(s2)

print("Set s1 and s2 combined with only unique values: ", s3)
```

```python
# 8) Write a Python program to find all elements that are common to both sets.

s1 = {1, 2, 3, 4} 
s2 = {3, 4, 5, 6}

s3 = s1.intersection(s2)

print("Common Values between s1 and s2: ", s3)
```

```python
# 9) Write a Python program to find all elements that are in either Set A or Set B, but not in both.

s1 = {1, 2, 3, 4} 
s2 = {3, 4, 5, 6}

s3 = s1.symmetric_difference(s2)

print("Elements in either set but not both:", s3)
```
