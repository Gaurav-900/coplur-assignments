```python
# 1) Write a Python program to create a Rectangle class with length and width as instance attributes, and two methods: area() and perimeter().

class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)


r = Rectangle(10, 5)
print("Area:", r.area())
print("Perimeter:", r.perimeter())
```

```python
# 2) Write a Python program to create a Circle class with radius as instance attribute, and methods diameter(), area() and circumference().

class Circle:
    def __init__(self, radius):
        self.radius = radius

    def diameter(self):
        return 2 * self.radius

    def area(self):
        return 3.14 * self.radius * self.radius

    def circumference(self):
        return 2 * 3.14 * self.radius


c = Circle(7)
print("Diameter:", c.diameter())
print("Area:", c.area())
print("Circumference:", c.circumference())
```

```python
# 3) Write a Python program to create a Product class with name, price and quantity. Add a method total_value().

class Product:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def total_value(self):
        return self.price * self.quantity


p = Product("Book", 250, 4)
print("Total Value:", p.total_value())
```

```python
# 4) Write a Python program to create a BankAccount class with deposit() and withdraw() methods.

class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance = self.balance + amount
        print("Balance:", self.balance)

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance = self.balance - amount
            print("Balance:", self.balance)
        else:
            print("Insufficient Balance")


b = BankAccount(5000)
b.deposit(1000)
b.withdraw(2000)
b.withdraw(5000)
```

```python
# 5) Write a Python program to create a User class with username and password. Add check_password() method.

class User:
    def __init__(self, username, password):
        self.username = username
        self.password = password

    def check_password(self, input_password):
        if input_password == self.password:
            return True
        else:
            return False


u = User("gaurav", "python123")

print(u.check_password("python123"))
print(u.check_password("hello"))
```
