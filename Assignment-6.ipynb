```python
# 1. Write a Python function to find the maximum of three numbers.

def maxn(a, b,c):
    if a > b and a > c:
        print(f"A: {a} is the Biggest")
    elif b > a and b > c:
        print(f"B: {b} is the Biggest")
    if c > b and c > a:
        print(f"C: {c} is the Biggest")

maxn(46, 34, 29)
print('')
maxn(15, 20, 12)
print('')
maxn(25, 38, 46)
```

```python
# 2. Write a Python function that takes a list and returns a new list with distinct elements from the first list.


def uni(l1):

    l = len(l1)
    l2= []
    for i in range(l):
        if l1[i] in l2:
            pass
        else:
            l2.append(l1[i])
    print("List before removing distinct elements: ", l1)
    return l2
    

lst = [3, 4, 6, 5, 3, 2, 4, 5, 6]
result = uni(lst)
print(f"List After removing distinct elements: {result}")
print("")
lst = [12, 14, 12, 13, 17, 14, 18, 14]
result = uni(lst)
print(f"List After removing distinct elements: {result}")
```

```python
# 3. Write a Python function to multiply all the numbers in a list.

def multi(lst):
    print("List: ", lst)
    mul = 1
    for l in lst:
        mul = mul * l
    return mul

    
l = [4, 5, 8]
result = multi(l)
print(f"Multiplication of all Numbers in the list is: {result}")

l = [2, 7, 3]
result = multi(l)
print(f"Multiplication of all Numbers in the list is: {result}")
```

```python
# 4. Write a Python function to calculate the factorial of a number (a non-negative integer). The function accepts the number as an argument.

def fac(n):
    if n < 0:
        print(f"Invalid!, {n} is a negative integer")
    else:
        fact = 1
        for i in range(n, 1, -1):
            print(i)
            fact = fact * i
        print(f"Factorial of {n} is: {fact}")

fac(5)
fac(-7)
```

```python
#  5. Write a Python program to reverse a string.

def rev(s1):
    print("String before reverse: ", s1)
    s2 = s1[::-1]
    print("String After reverse",s2)

s = str(input('Enter a string'))
rev(s)
```

```python
# 8. Write a Python function that takes a number as a parameter and checks whether the number is prime or not.

def check(n):
    if n % 2 == 0:
        print(n, ' is a prime number')
    else:
        print(n, ' is not  a prime number')

check(12)
check(15)
```

```python
# 9. Write a Python function that accepts a string and counts the number of upper and lower case letters. 

def counter(st):
    lower = 0
    upper = 0
    print("String:", st)
    for i in st:
        if i.isupper() == True:
            upper = upper + 1
        elif i == " ":
            pass
        else:
            lower = lower + 1
    print('LowerCase Count:', lower)
    print( "UpperCase Count:", upper)

counter("Python")
print("")
counter("Hello World")
    
```
