```python
# 1) Practise API calling. Complete program to get live stock data using Alpha Vantage API.

import requests

api_key = "BAEO2AO3M7YMLVWE "

symbol = input("Enter Stock Symbol: ").upper()

url = "https://www.alphavantage.co/query"

parameters = {
    "function": "GLOBAL_QUOTE",
    "symbol": symbol,
    "apikey": api_key
}

try:
    response = requests.get(url, params=parameters)
    data = response.json()

    price = data["Global Quote"]["05. price"]

    print("Stock Symbol:", symbol)
    print("Current Price:", price)

except:
    print("Error: Unable to fetch stock data.")
```

```python
# 2) Add Exception methods in your Assignment 13 programs. Example with Temperature class

class Temperature:
    def __init__(self, celsius):
        self.celsius = celsius

    def to_fahrenheit(self):
        try:
            return (self.celsius * 9 / 5) + 32
        except:
            print("Invalid Input")

    def to_kelvin(self):
        try:
            return self.celsius + 273.15
        except:
            print("Invalid Input")


try:
    t = Temperature(25)

    print("Fahrenheit:", t.to_fahrenheit())
    print("Kelvin:", t.to_kelvin())

except Exception as e:
    print("Error:", e)
```
