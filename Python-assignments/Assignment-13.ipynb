```python
# 1) Write a Python program to create a Temperature class that converts Celsius to Fahrenheit and Kelvin.

class Temperature:
    def __init__(self, celsius):
        self.celsius = celsius

    def to_fahrenheit(self):
        return (self.celsius * 9 / 5) + 32

    def to_kelvin(self):
        return self.celsius + 273.15


t = Temperature(30)

print("Fahrenheit:", t.to_fahrenheit())
print("Kelvin:", t.to_kelvin())
```

```python
# 2) Write a Python program to create an interest calculation class. Add methods to find simple interest and compound interest.

class Interest:
    def __init__(self, p, r, t):
        self.p = p
        self.r = r
        self.t = t

    def simple_interest(self):
        return (self.p * self.r * self.t) / 100

    def compound_interest(self):
        amount = self.p * (1 + self.r / 100) ** self.t
        return amount - self.p


i = Interest(10000, 10, 2)

print("Simple Interest:", i.simple_interest())
print("Compound Interest:", round(i.compound_interest(), 2))
```

```python
# 3) Write a Python program to create a CoffeeMachine class.

class CoffeeMachine:
    def __init__(self, water, coffee, milk):
        self.water = water
        self.coffee = coffee
        self.milk = milk

    def make_latte(self):
        if self.water >= 200 and self.coffee >= 20 and self.milk >= 150:
            self.water -= 200
            self.coffee -= 20
            self.milk -= 150
            print("Latte is ready.")
        else:
            print("Not enough resources.")


machine = CoffeeMachine(300, 100, 200)

machine.make_latte()
machine.make_latte()
```

```python
# 4) Write a Python program to create a Vehicle parent class and Bus child class.

class Vehicle:
    def __init__(self, name, max_speed):
        self.name = name
        self.max_speed = max_speed

    def display(self):
        print("Vehicle:", self.name)
        print("Max Speed:", self.max_speed)


class Bus(Vehicle):
    def __init__(self, name, max_speed, seats):
        super().__init__(name, max_speed)
        self.seats = seats

    def reservation(self):
        print("Reservation done for today.")

    def display(self):
        super().display()
        print("Seats:", self.seats)


b = Bus("Travel Bus", 100, 50)

b.display()
b.reservation()
```
